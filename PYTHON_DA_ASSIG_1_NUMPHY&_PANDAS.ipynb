```python
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
print(type(temperatures_w1))


```

```python
print("Shape            :", temperatures_w1.shape)
print("Data type        :", temperatures_w1.dtype)
print("Number of items  :", temperatures_w1.size)
print("Dimensions       :", temperatures_w1.ndim)
```

```python
temperatures_f = (temperatures_w1 * 9/5) + 32
print(np.round(temperatures_f, 2))
```

```python
print("Maximum:", temperatures_w1.max())
print("Minimum:", temperatures_w1.min())
print("Mean   :", round(temperatures_w1.mean(), 2))
```

```python
first_three = temperatures_w1[0:3]     # days 1, 2, 3
weekend     = temperatures_w1[-2:]     # last two days
middle_three = temperatures_w1[2:5]    # days 3, 4, 5

print("First three days :", first_three)
print("Weekend          :", weekend)
print("Middle three days:", middle_three)
```

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],   # Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]    # Week 2
])
print(temperatures)
```

```python
print("Shape           :", temperatures.shape)
print("Data type       :", temperatures.dtype)
print("Total elements  :", temperatures.size)
```

```python
week1 = temperatures[0]
week2 = temperatures[1]
print("Week 1:", week1)
print("Week 2:", week2)

weekends = temperatures[:, -2:]      # all rows, last two columns
print("Weekends of both weeks:")
print(weekends)
print("Week 1 weekend:", temperatures[0, -2:])
print("Week 2 weekend:", temperatures[1, -2:])
```

```python
import pandas as pd

marks = pd.Series([95, 92, 89, 85, 80],
                  index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)
```

```python
print("1st rank mark (position 0):", marks.iloc[0])
print("1st rank mark (label)       :", marks['Rank1'])
```

```python
top3 = marks.loc['Rank1':'Rank3']
print(top3)
```

```python
print("3rd rank mark:", marks.iloc[2])
```

```python
mask = marks > 90
print(mask)
print()
above_90 = marks[mask]
print(above_90)
print("Ranks:", list(above_90.index))
```

```python
marks['Rank1'] = 100
print(marks)
```

```python
marks = marks.drop('Rank5')
print(marks)
```

```python
cgpa = marks / 10
print(cgpa)
```

```python
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West',
               'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
print(transactions)
```

```python
print(transactions.head())
```

```python
print(transactions.tail())
```

```python
print("Shape (rows, columns):", transactions.shape)
print("Column names          :", list(transactions.columns))
print()
print("Data types:")
print(transactions.dtypes)
```

```python
north_high = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
print(north_high)
```

```python
print(transactions['ProductCategory'].value_counts())
```

```python
print(transactions['Region'].unique())
print("Number of unique regions:", transactions['Region'].nunique())
```

```python
print(transactions.groupby('Region')['Amount'].mean())
```

```python
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print(transactions[transactions['TransactionID'] == 102])
```

```python
transactions['Discount'] = transactions['Amount'] * 0.10
print(transactions)
```

```python
transactions = transactions[transactions['TransactionID'] != 109]
print(transactions)
```
